```python
import pandas as pd
import numpy as np
import os
```

```python
student_file = "../data/processed/cleaned_student_dataset.csv"
reddit_file =  "../data/processed/cleaned_reddit_dataset.csv"

student_df = pd.read_csv(student_file)
reddit_df = pd.read_csv(reddit_file)
```

# MERGED HYBRID DATASET (STUDENT DATA AND REDDIT DATA)

```python
# Prepare the Reddit dataset
reddit_clean = reddit_df[
    reddit_df["clean_text"].str.strip() != "#name?"
][['clean_text', 'binary_label']].copy()

reddit_clean['source'] = 'Reddit'

# Prepare the Student dataset
student_clean = student_df[['serialized_text', 'Depression']].copy()

student_clean = student_clean.rename(columns={
    'serialized_text': 'clean_text',
    'Depression': 'binary_label'
})

student_clean['source'] = 'Student'

# Combine both datasets
combined_df = pd.concat(
    [reddit_clean, student_clean],
    ignore_index=True
)

# Remove missing and empty text records
combined_df = combined_df.dropna(
    subset=['clean_text', 'binary_label']
)

combined_df['clean_text'] = combined_df['clean_text'].astype(str).str.strip()

combined_df = combined_df[combined_df['clean_text'] != ''].copy()

# Shuffle reproducibly
combined_df = combined_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

# Save the combined dataset
combined_output_path = ("../data/processed/combined_hybrid_student_reddit_data.csv")

combined_df.to_csv(combined_output_path, index=False)

# Preview the result
combined_df.head()
```
